La simulación conserva la escala lineal real para mostrar cuánto se separan las familias de complejidad cuando aumenta $n$.

## Instrucciones

1. Elige un valor objetivo entre $1$ y $1000$.
2. Selecciona la vista y la velocidad.
3. Pulsa **Ejecutar** para observar un crecimiento continuo o **Reiniciar** para volver al origen.

En la escala lineal, las curvas pequeñas pueden quedar prácticamente sobre cero. Esto no es un error visual: expresa la diferencia real de magnitud frente a las funciones dominantes.

```python
#@title Simulación interactiva del crecimiento de funciones { display-mode: "form" }

import os
import time
import urllib.request
from pathlib import Path

SIMULATION_NAME = "growth_comparison"
_bootstrap_url = "https://raw.githubusercontent.com/Notas-a-Mano-serie-de-libros/3_notas-a-mano-sobre-analisis-de-complejidad-computacional/main/simulaciones/capitulo2/runtime/colab_bootstrap.py"
if os.environ.get("COLAB_RELEASE_TAG"):
    _bootstrap_source = urllib.request.urlopen(
        f"{_bootstrap_url}?runtime={time.time_ns()}"
    ).read()
else:
    _bootstrap_path = next(
        (
            candidate
            for base in (Path.cwd(), *Path.cwd().parents)
            for candidate in (
                base / "simulaciones" / "capitulo2" / "runtime" / "colab_bootstrap.py",
                base / "colab_bootstrap.py",
            )
            if candidate.exists()
        ),
        None,
    )
    if _bootstrap_path is not None:
        _bootstrap_source = _bootstrap_path.read_bytes()
    else:
        _bootstrap_source = urllib.request.urlopen(
            f"{_bootstrap_url}?runtime={time.time_ns()}"
        ).read()
exec(_bootstrap_source)
```
