# Comparación del crecimiento de funciones · simulación ejecutable

La simulación conserva la escala lineal real para mostrar cuánto se separan las familias de complejidad cuando aumenta $n$.

## Instrucciones

1. Elige un valor objetivo entre $1$ y $1000$.
2. Selecciona la velocidad, la escala y las funciones que deseas comparar.
3. Pulsa **Reproducir** para comenzar, **Pausar/Continuar** para controlar la animación, **Finalizar** para saltar al resultado o **Reiniciar** para volver a $n=0$ con un valor final predeterminado de $100$.

El eje horizontal conserva siempre el límite de $n$ configurado y el valor actual se actualiza durante la animación. En la escala lineal, las curvas pequeñas pueden quedar prácticamente sobre cero porque se respetan las diferencias reales de magnitud.


```python
#@title Simulación interactiva del crecimiento de funciones { display-mode: "form" }

import base64
import gzip
import importlib
import os
import sys
import time
import urllib.request
from pathlib import Path

_bootstrap_url = "https://raw.githubusercontent.com/Notas-a-Mano-serie-de-libros/3_notas-a-mano-sobre-analisis-de-complejidad-computacional/8e4db91d4f199230f179a8b9ad49d27e93f26a6d/simulaciones/capitulo2/runtime/colab_bootstrap.py"
if os.environ.get("COLAB_RELEASE_TAG") or sys.platform == "emscripten":
    # Colab y JupyterLite ejecutan el runtime versionado que viaja en el notebook.
    if os.environ.get("COLAB_RELEASE_TAG"):
        globals().pop("SIMULATION_NAME", None)
        _bootstrap_source = urllib.request.urlopen(
            f"{_bootstrap_url}?runtime={time.time_ns()}"
        ).read()
        exec(_bootstrap_source)
    _runtime_payload = ''
    _project_root = next(
        (base for base in (Path.cwd(), *Path.cwd().parents) if (base / "common").is_dir()),
        Path.cwd(),
    )
    if str(_project_root) not in sys.path:
        sys.path.insert(0, str(_project_root))
    _runtime_path = _project_root / "simulaciones/capitulo2/runtime/growth_comparison.py"
    _runtime_path.parent.mkdir(parents=True, exist_ok=True)
    _runtime_path.write_bytes(gzip.decompress(base64.b64decode(_runtime_payload)))
    importlib.invalidate_caches()
    sys.modules.pop("simulaciones.capitulo2.runtime.growth_comparison", None)
    from IPython.display import clear_output
    clear_output(wait=True)
    importlib.import_module("simulaciones.capitulo2.runtime.growth_comparison").run_app()
else:
    SIMULATION_NAME = "growth_comparison"
    _bootstrap_path = next(
        (
            candidate
            for base in (Path.cwd(), *Path.cwd().parents)
            for candidate in (
                base / "simulaciones" / "capitulo2" / "runtime" / "colab_bootstrap.py",
                base / "colab_bootstrap.py",
            )
            if candidate.exists()
        ),
        None,
    )
    if _bootstrap_path is not None:
        _bootstrap_source = _bootstrap_path.read_bytes()
    else:
        _bootstrap_source = urllib.request.urlopen(
            f"{_bootstrap_url}?runtime={time.time_ns()}"
        ).read()
    exec(_bootstrap_source)
```
