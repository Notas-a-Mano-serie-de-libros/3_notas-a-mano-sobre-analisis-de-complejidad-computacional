# Ejemplo 9: Fibonacci iterativo

Este ejemplo calcula la sucesión de Fibonacci mediante un ciclo. La implementación conserva la estructura de la solución original y utiliza `numpy.int32` para reproducir en Python los enteros de 32 bits de Java.

## Código analizado

```python
import numpy as np

def fibonacci_iterativo(n):
    if n <= 1:
        return np.int32(n)

    a, b = np.int32(0), np.int32(1)
    with np.errstate(over="ignore"):
        for i in range(2, n + 1):
            c = a + b
            a = b
            b = c
    return b
```

## Análisis esperado

### Complejidad temporal

El ciclo realiza una cantidad de iteraciones proporcional a $n$. Cada iteración opera sobre enteros de 32 bits y tiene costo constante; por tanto, $T(n)\in O(n)$.

### Complejidad espacial

El algoritmo mantiene únicamente las variables `a`, `b`, `c` e `i`. Su cantidad y tamaño no dependen de $n$; por tanto, $S(n)\in O(1)$.

> Como ocurre con `int` en Java, el resultado se desborda cuando deja de caber en 32 bits. `np.errstate` solo oculta la advertencia de NumPy; no modifica el cálculo ni su costo. El ejemplo 9 estudia por separado Fibonacci con enteros de precisión arbitraria.
