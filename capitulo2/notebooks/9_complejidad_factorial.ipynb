```python
def contar_permutaciones(lista):
    if len(lista) <= 1:
        return 1

    total = 0
    for indice in range(len(lista)):
        restante = lista[:indice] + lista[indice + 1:]
        total += contar_permutaciones(restante)

    return total
```
