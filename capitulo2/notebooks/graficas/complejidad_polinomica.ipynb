```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
plt.style.use('default')
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['savefig.facecolor'] = 'white'
plt.rcParams['savefig.edgecolor'] = 'white'

import sys

plt.rcParams['font.family'] = 'STIXGeneral'
plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['axes.formatter.use_mathtext'] = True
plt.rcParams['axes.titlesize'] = 15
plt.rcParams['axes.labelsize'] = 13
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10

try:
    from common.graphics import graphics_dir
except ModuleNotFoundError:
    PROJECT_ROOT = next(
        (base for base in (Path.cwd(), *Path.cwd().parents) if (base / 'common' / 'graphics.py').is_file()),
        None,
    )
    if PROJECT_ROOT is not None:
        sys.path.insert(0, str(PROJECT_ROOT))
        from common.graphics import graphics_dir
    else:
        def graphics_dir(chapter, *parts):
            destination = Path.cwd() / chapter / 'images' / 'generadas' / Path(*parts)
            destination.mkdir(parents=True, exist_ok=True)
            return destination

GRAPHICS_DIR = graphics_dir('capitulo2', 'general')

def aplicar_estilo_polinomico(axis, titulo):
    axis.set_xlabel(r'$\mathrm{Tamaño\ de\ la\ entrada}\ (n)$')
    axis.set_ylabel(r'$\mathrm{Función\ de\ complejidad\ teórica}$')
    axis.set_title(titulo)
    axis.xaxis.set_label_coords(0.5, -0.12)
    axis.yaxis.set_label_coords(-0.075, 0.5)
    axis.title.set_position((0.5, 1.02))
    x_formatter = plt.ScalarFormatter(useMathText=True)
    x_formatter.set_scientific(False)
    x_formatter.set_useOffset(False)
    axis.xaxis.set_major_formatter(x_formatter)
    y_formatter = plt.ScalarFormatter(useMathText=True)
    y_formatter.set_scientific(True)
    y_formatter.set_powerlimits((0, 0))
    axis.yaxis.set_major_formatter(y_formatter)
    axis.yaxis.get_offset_text().set_fontfamily('STIXGeneral')
    axis.tick_params(axis='both', labelsize=10)
    for tick_label in (*axis.get_xticklabels(), *axis.get_yticklabels()):
        tick_label.set_fontfamily('STIXGeneral')
    axis.grid(True, color='#CFD8DC', linestyle='-', linewidth=0.6, alpha=0.55)
    for spine in axis.spines.values():
        spine.set_color('#000000')
        spine.set_linewidth(0.8)

plt.rcParams['figure.dpi'] = 500
plt.rcParams['savefig.dpi'] = 500
```

```python

k_max_1 = 4
ylim_1 = 4
valores_k_1 = np.arange(0, k_max_1 + 1, 1)
tamanos_problema_1 = np.linspace(1, 10, 10**ylim_1)

fig1, ax1 = plt.subplots(1, 1, figsize=(8, 4))

ax1.set_ylim([0, 10])
ax1.set_xlim([1, 10.6])

annotate_positions_1 = {
    4: (1.2, 8),
    3: (2.0, 8),
    2: (2.9, 8),
    1: (8.4, 8),
    0: (10, 1)
}

for k in valores_k_1:
    complejidad_teorica = tamanos_problema_1 ** k
    line, = ax1.plot(tamanos_problema_1, complejidad_teorica, label=f'$n^{{{k}}}$')

    if k in annotate_positions_1:
        ax1.annotate(f'$n^{{{k}}}$', xy=annotate_positions_1[k],
                     xytext=(5, 0), textcoords='offset points', fontsize=14, color=line.get_color())

aplicar_estilo_polinomico(ax1, rf'$C(n)=n^k$ para $k \in [0, {ylim_1}]$')

plt.tight_layout()
fig1.savefig(GRAPHICS_DIR / 'complejidad_teorica_polinomica1.png', bbox_inches="tight", pad_inches=.05)
plt.show()
```

```python

k_max_2 = 10
ylim_2 = 6
valores_k_2 = np.arange(0, k_max_2 + 1, 1)
tamanos_problema_2 = np.linspace(1, 10, 10**ylim_2)

fig2, ax2 = plt.subplots(1, 1, figsize=(8, 4))

ax2.set_ylim([-10**5, 10**ylim_2])
ax2.set_xlim([2, 10.8])

valor_referencia_2 = 0.5 * 10**ylim_2

etiqueta_n_0_k_añadida = False

for k in valores_k_2:
    complejidad_teorica = tamanos_problema_2 ** k
    line, = ax2.plot(tamanos_problema_2, complejidad_teorica, label=f'$n^{{{k}}}$')

    if k >= ylim_2:
        position_index = np.abs(complejidad_teorica - valor_referencia_2).argmin()
        ax2.annotate(f'$n^{{{k}}}$', xy=(tamanos_problema_2[position_index]+.05, complejidad_teorica[position_index]),
                     xytext=(5, 0), textcoords='offset points', fontsize=14, color=line.get_color())
    elif k == ylim_2 - 1:
        position_index = -1
        ax2.annotate(f'$n^{ylim_2-1}$', xy=(tamanos_problema_2[position_index], complejidad_teorica[position_index]),
                     xytext=(5, 0), textcoords='offset points', fontsize=14, color=line.get_color())
    elif not etiqueta_n_0_k_añadida:
        position_index = -1
        ax2.annotate(f'$n^{{{[0, ylim_2-2]}}}$', xy=(tamanos_problema_2[position_index], complejidad_teorica[position_index]),
                     xytext=(5, 0), textcoords='offset points', fontsize=14, color='black')
        etiqueta_n_0_k_añadida = True

aplicar_estilo_polinomico(ax2, rf'$C(n)=n^k$ para $k \in [0, {k_max_2}]$')
ax2.ticklabel_format(axis='y', style='sci', scilimits=(0, 0))

plt.tight_layout()
fig2.savefig(GRAPHICS_DIR / 'complejidad_teorica_polinomica2.png', bbox_inches="tight", pad_inches=.05)
plt.show()
```
