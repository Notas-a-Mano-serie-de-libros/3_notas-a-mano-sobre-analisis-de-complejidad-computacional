```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
from pathlib import Path
plt.style.use('default')
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['savefig.facecolor'] = 'white'
plt.rcParams['savefig.edgecolor'] = 'white'
plt.rcParams['font.family'] = 'STIXGeneral'
plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['axes.formatter.use_mathtext'] = True

plt.rcParams['figure.dpi'] = 500
plt.rcParams['savefig.dpi'] = 500
plt.subplots_adjust(left=0, right=1, top=1, bottom=0)

n_values = np.logspace(0, 4, num=5, dtype=int)
```

```python

fig, ax2 = plt.subplots(1, 1, figsize=(8, 5))

x = [1e0, 1e1, 1e2, 1e3, 1e4]
y = [0.5, 0.8, 24, 295, 3004]

log_x = np.log10(x)
log_y = np.log10(y)
interp_func_2 = interp1d(log_x, log_y, fill_value="extrapolate")
log_x_for_y_17_2 = interp_func_2(np.log10(17))
x_for_y_17_2 = 10 ** log_x_for_y_17_2

ax2.set_yscale('log')
ax2.set_xscale('log')
ax2.set_ylim(1e-1, 1e4)

yticks = [1E-1, 1e0, 1e1, 1e2, 1e3, 1e4]
ytick_labels = [r'$10^{-10}$', r'$10^{0}$', r'$10^{10}$', r'$10^{100}$', r'$10^{1000}$', r'$10^{10000}$']
plt.yticks(yticks, ytick_labels)

ax2.set_xlabel(r'$\mathrm{Tamaño\ de\ la\ entrada}\ (n)$')
ax2.set_ylabel(r'$\mathrm{Tiempo\ de\ ejecución\ teórico}\ [s]$')
ax2.set_title(r'$\mathrm{Comportamiento\ de\ la\ función}\ T(n)=2^n$')
ax2.plot(x, y, marker='o', linestyle='-', color='b', label=r'$\mathrm{Complejidad\ temporal\ teórica}$')

ax2.scatter(76, 17, color='red', marker='*', s=100, label=r'$\mathrm{Edad\ estimada\ del\ Universo}:\ 4.17\cdot10^{17}\ [s]$', zorder=5)
ax2.grid(True, which="both", ls="--")

ax2.legend(loc='upper left')

plt.show()
import sys
from pathlib import Path
PROJECT_ROOT = next(base for base in (Path.cwd(), *Path.cwd().parents) if (base / 'common' / 'graphics.py').is_file())
if str(PROJECT_ROOT) not in sys.path: sys.path.insert(0, str(PROJECT_ROOT))
from common.graphics import graphics_dir
GRAPHICS_DIR = graphics_dir("capitulo2", "temporal")
fig.savefig(GRAPHICS_DIR / 'complejidad_exponencial_1.png', bbox_inches="tight", pad_inches=.05)
```
