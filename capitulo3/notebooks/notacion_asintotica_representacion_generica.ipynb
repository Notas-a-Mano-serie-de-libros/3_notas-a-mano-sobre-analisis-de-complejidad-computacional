```python
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path
PROJECT_ROOT = next(base for base in (Path.cwd(), *Path.cwd().parents) if (base / 'common' / 'graphics.py').is_file())
if str(PROJECT_ROOT) not in sys.path: sys.path.insert(0, str(PROJECT_ROOT))
from common.graphics import graphics_dir

GRAPH_DIR = graphics_dir('capitulo3', 'referencias')

COLOR_C = '#1565C0'
COLOR_SUP = '#C62828'
COLOR_INF = '#EF6C00'
COLOR_N0 = '#2E7D32'
COLOR_GRID = '#CFD8DC'
COLOR_THETA_FILL = '#78909C'
GENERIC_FIGSIZE = (8, 3)

def style_generic_axis(ax, title):
    ax.set_xlabel(r'$\mathrm{Tamaño\ de\ la\ entrada}\ (n)$', fontsize=10, labelpad=7)
    ax.set_ylabel(r'$\mathrm{Complejidad\ general}$', fontsize=10, labelpad=7)
    ax.set_title(title, fontsize=12, fontweight='normal', pad=8)
    ax.set_xticks(np.linspace(*ax.get_xlim(), 6))
    ax.set_yticks(np.linspace(*ax.get_ylim(), 5))
    ax.set_xticklabels([])
    ax.set_yticklabels([])
    ax.tick_params(axis='both', length=0)
    ax.set_axisbelow(True)
    ax.grid(True)
    for side in ('left', 'bottom', 'top', 'right'):
        ax.spines[side].set_visible(True)
        ax.spines[side].set_color('#000000')
        ax.spines[side].set_linewidth(0.8)
    ax.legend(loc='upper left', fontsize=9, frameon=True, framealpha=0.88,
              facecolor='white', edgecolor='#E0E0E0', labelspacing=0.35,
              handlelength=2.2, handletextpad=0.7, borderpad=0.5)
    ax.annotate(r'$n_0$', xy=(n0, ax.get_ylim()[1]), xycoords='data',
                xytext=(6, -8), textcoords='offset points', ha='left', va='top',
                color=COLOR_N0, fontsize=10)
    ax.annotate(r'$n \to \infty$', xy=(1, 0), xycoords='axes fraction', fontsize=13,
                color='#000000', xytext=(-10, 7), textcoords='offset points',
                ha='right', va='bottom', zorder=5)

plt.style.use('default')
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['savefig.facecolor'] = 'white'
plt.rcParams['savefig.edgecolor'] = 'white'
plt.rcParams['font.family'] = 'STIXGeneral'
plt.rcParams['mathtext.fontset'] = 'stix'

plt.rcParams['figure.dpi'] = 500
plt.rcParams['savefig.dpi'] = 500

plt.subplots_adjust(left=0, right=1, top=1, bottom=0)
```

```python
#@title Notacion_O

# Definir funciones arbitrarias C(n) y g(n)
def C(n):
    return 8.73 - 7.96*n + 3.21*n**2 - 0.46*n**3 + 0.023*n**4

def g(n):
    return 3.38 - 0.53*n - 0.13*n**2 + 0.12*n**3 - 0.0086*n**4

# Constante c
c = 2

# Punto n_0
n0 = 4.1

# Rango de valores de n
n = np.linspace(.01, 8, 1000)

# Calcular los valores de las funciones
f_values = C(n)
g_values = g(n)

# Crear la gráfica
fig, ax = plt.subplots(figsize=GENERIC_FIGSIZE)

# Configurar etiquetas
label_text = r'$c \cdot g(n)$'
title_text = r'Comparación entre $C(n)$ y $c \cdot g(n)$ en notación $O$'

# Graficar las funciones
ax.plot(n, f_values, label=r'$C(n)$', color=COLOR_C, linewidth=2.2,
        solid_capstyle='round', solid_joinstyle='round', zorder=3)
ax.plot(n, g_values, label=label_text, color=COLOR_SUP, linewidth=2.2,
        solid_capstyle='round', solid_joinstyle='round', zorder=3)
ax.axvline(x=n0, color=COLOR_N0, linestyle='--', linewidth=1.5,
           label=r'$n_0$', zorder=4)

# Rellenar el área entre las funciones después de n_0
n_fill = n[n >= n0]
f_fill = f_values[n >= n0]
g_fill = g_values[n >= n0]
ax.fill_between(n_fill, f_fill, g_fill, color=COLOR_SUP, alpha=0.10,
                edgecolor='none', linewidth=0, zorder=2)

# Aplicar el estilo visual compartido
style_generic_axis(ax, title_text)

# Guardar la imagen
fig.savefig(GRAPH_DIR / 'comparacion_o_generica.png', dpi=500, bbox_inches="tight", pad_inches=.05)
plt.show()
```

```python
# Definir funciones arbitrarias C(n) y g(n), ahora adaptadas para notación Omega
def C(n):
    return 10 + 3*np.sin(5*n) + 0.2*n**3

def g(n):
    return 5 + 2*np.cos(3*n) + 0.1*n**3

# Rango de valores de n
n = np.linspace(.01, 8, 1000)

# Calcular valores de funciones
f_values = C(n)
g_values = 1.2 * g(n)

# Crear gráfica
fig, ax = plt.subplots(figsize=GENERIC_FIGSIZE)

# Configurar etiquetas
label_text = r'$c \cdot g(n)$'
title_text = r'Comparación entre $C(n)$ y $c \cdot g(n)$ en notación $\Omega$'

n0 = 2.3

# Graficar funciones
ax.plot(n, f_values, label=r'$C(n)$', color=COLOR_C, linewidth=2.2,
        solid_capstyle='round', solid_joinstyle='round', zorder=3)
ax.plot(n, g_values, label=label_text, color=COLOR_INF, linewidth=2.2,
        solid_capstyle='round', solid_joinstyle='round', zorder=3)
ax.axvline(x=n0, color=COLOR_N0, linestyle='--', linewidth=1.5,
           label=r'$n_0$', zorder=4)

# Rellenar área entre funciones después de n_0 en azul
n_fill = n[n >= n0]
f_fill = f_values[n >= n0]
g_fill = g_values[n >= n0]
ax.fill_between(n_fill, f_fill, g_fill, where=(f_fill >= g_fill),
                color=COLOR_C, alpha=0.10, edgecolor='none', linewidth=0, zorder=2)

# Aplicar el estilo visual compartido
style_generic_axis(ax, title_text)

# Guardar imagen
fig.savefig(GRAPH_DIR / 'comparacion_omega_generica.png', dpi=500, bbox_inches="tight", pad_inches=.05)
plt.show()
```

```python
# Definir funciones arbitrarias
def f_sup(n):
    return 10 + 2 * np.sin(3 * n) + 0.55 * n**3

def f_med(n):
    return 8 + 1.5 * np.cos(2 * n) + 0.5 * n**3

def f_inf(n):
    return 6 + 2 * np.sin(4 * n + 1) + 0.45 * n**3

# Rango de valores de n
n = np.linspace(1.195, 3, 1000)

# Evaluar funciones
sup_vals = f_sup(n)
med_vals = f_med(n)
inf_vals = f_inf(n)

# Crear la gráfica
fig, ax = plt.subplots(figsize=GENERIC_FIGSIZE)

n0 = 1.95

# Graficar las tres funciones con etiquetas en notación matemática
ax.plot(n, med_vals, color=COLOR_C, linewidth=2.2, label=r'$C(n)$',
        solid_capstyle='round', solid_joinstyle='round', zorder=3)
ax.plot(n, sup_vals, color=COLOR_SUP, linewidth=2.2, label=r'$c_2 \cdot g(n)$',
        solid_capstyle='round', solid_joinstyle='round', zorder=3)
ax.plot(n, inf_vals, color=COLOR_INF, linewidth=2.2, label=r'$c_1 \cdot g(n)$',
        solid_capstyle='round', solid_joinstyle='round', zorder=3)
ax.axvline(x=n0, color=COLOR_N0, linestyle='--', linewidth=1.5,
           label=r'$n_0$', zorder=4)

# Rellenar el área entre la función superior e inferior desde n = 2

mask = n >= n0
ax.fill_between(n[mask], inf_vals[mask], sup_vals[mask], color=COLOR_THETA_FILL,
                alpha=0.10, edgecolor='none', linewidth=0, zorder=2)

# Aplicar el estilo visual compartido
title_text = r'Comparación entre $C(n)$ y las referencias en notación $\Theta$'
style_generic_axis(ax, title_text)

# Ajustar, guardar y mostrar la gráfica
fig.savefig(GRAPH_DIR / 'comparacion_theta_generica.png', dpi=500, bbox_inches="tight", pad_inches=.05)
plt.show()
```
