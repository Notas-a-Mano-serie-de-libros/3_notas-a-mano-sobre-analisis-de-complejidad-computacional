```python
import matplotlib.pyplot as plt
from matplotlib import ticker
from pathlib import Path

plt.style.use('default')
plt.rcParams['figure.dpi']        = 500
plt.rcParams['savefig.dpi']       = 500
plt.rcParams['figure.facecolor']  = 'white'
plt.rcParams['axes.facecolor']    = 'white'
plt.rcParams['savefig.facecolor'] = 'white'
plt.rcParams['savefig.edgecolor'] = 'white'
plt.rcParams['font.family'] = 'STIXGeneral'
plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['axes.formatter.use_mathtext'] = True
plt.rcParams['axes.titleweight'] = 'semibold'
WORKING_DIR = Path.cwd()
CHAPTER_DIR = (WORKING_DIR / 'capitulo6' if (WORKING_DIR / 'capitulo6').is_dir()
               else WORKING_DIR if WORKING_DIR.name == 'capitulo6'
               else WORKING_DIR.parent if WORKING_DIR.parent.name == 'capitulo6'
               else WORKING_DIR)
GRAPH_DIR = CHAPTER_DIR / 'graficas'
GRAPH_DIR.mkdir(parents=True, exist_ok=True)

def aplicar_funcion(n, funcion):
    valores = [n]
    while n > 1:
        n = funcion(n)
        valores.append(n)
    return valores

n_inicial = 100

valores_n1 = aplicar_funcion(n_inicial, lambda x: x - 1)
valores_n2 = aplicar_funcion(n_inicial, lambda x: x - 2)

pasos_n1 = len(valores_n1) - 1
pasos_n2 = len(valores_n2) - 1

fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(valores_n1, color='#2563eb', linewidth=2.2,
        label=rf'$C(n-1)$ — {pasos_n1} pasos')
ax.plot(valores_n2, color='#dc2626', linewidth=2.2,
        linestyle='--', label=rf'$C(n-2)$ — {pasos_n2} pasos')

ax.set_xlabel(r'$\mathrm{Número\ de\ llamadas}$', fontsize=12)
ax.set_ylabel(r'$\mathrm{Tamaño\ del\ problema}\ (n)$', fontsize=12)
ax.set_title(rf'Decrecimiento desde $n = {n_inicial}$', fontsize=14, pad=12)
ax.xaxis.set_major_formatter(ticker.ScalarFormatter(useMathText=True))
ax.yaxis.set_major_formatter(ticker.ScalarFormatter(useMathText=True))
ax.tick_params(axis='both', labelsize=10, colors='#333333')
ax.legend(fontsize=10.5, frameon=True, framealpha=0.95,
          facecolor='white', edgecolor='#E0E0E0')
ax.grid(True, color='#CFD8DC', linestyle='-', linewidth=0.6, alpha=0.55)
for spine in ax.spines.values():
    spine.set_color('#000000')
    spine.set_linewidth(0.8)

plt.tight_layout()

fig.savefig(GRAPH_DIR / 'comparacion_llamados_recursivos.png', dpi=500,
            bbox_inches='tight', pad_inches=0.05)

plt.show()
```
