```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import ticker
from pathlib import Path

plt.style.use('default')
plt.rcParams['figure.dpi'] = 500
plt.rcParams['savefig.dpi'] = 500
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['savefig.facecolor'] = 'white'
plt.rcParams['savefig.edgecolor'] = 'white'
plt.rcParams['font.family'] = 'STIXGeneral'
plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['axes.formatter.use_mathtext'] = True
plt.rcParams['axes.titleweight'] = 'semibold'
WORKING_DIR = Path.cwd()
CHAPTER_DIR = (WORKING_DIR / 'capitulo6' if (WORKING_DIR / 'capitulo6').is_dir()
               else WORKING_DIR if WORKING_DIR.name == 'capitulo6'
               else WORKING_DIR.parent if WORKING_DIR.parent.name == 'capitulo6'
               else WORKING_DIR)
import sys
from pathlib import Path
PROJECT_ROOT = next(base for base in (Path.cwd(), *Path.cwd().parents) if (base / 'common' / 'graphics.py').is_file())
if str(PROJECT_ROOT) not in sys.path: sys.path.insert(0, str(PROJECT_ROOT))
from common.graphics import graphics_dir
GRAPH_DIR = graphics_dir('capitulo6', 'referencias')

n = np.linspace(1, 30, 500)

phi = (1 + np.sqrt(5)) / 2
exact    = phi**n                    # Θ(φⁿ) — solución exacta
upper    = 2**n                      # O(2ⁿ)  — cota superior
lower    = (np.sqrt(2))**n           # Ω(√2ⁿ) — cota inferior

fig, ax = plt.subplots(figsize=(8, 4))

ax.semilogy(n, exact, color='#2563eb', linewidth=2.5,
            label=r'$\Theta(\varphi^n)$ — solución exacta')
ax.semilogy(n, upper, color='#dc2626', linewidth=2,
            linestyle='--', label=r'$O(2^n)$ — cota superior')
ax.semilogy(n, lower, color='#16a34a', linewidth=2,
            linestyle='--', label=r'$\Omega((\sqrt{2})^n)$ — cota inferior')

ax.fill_between(n, lower, upper, alpha=0.07, color='#2563eb')

ax.set_xlabel(r'$n$', fontsize=12)
ax.set_ylabel(r'$T(n)$', fontsize=12)
ax.set_title('Fibonacci recursivo: cotas asintóticas', fontsize=14, pad=12)
ax.tick_params(axis='both', labelsize=10, colors='#333333')
ax.yaxis.set_major_formatter(ticker.LogFormatterMathtext())
ax.legend(fontsize=10.5, frameon=True, framealpha=0.95,
          facecolor='white', edgecolor='#E0E0E0')
ax.grid(True)
for spine in ax.spines.values():
    spine.set_color('#000000')
    spine.set_linewidth(0.8)

plt.tight_layout()

fig.savefig(GRAPH_DIR / 'comparacion_fibonacci.png', dpi=500,
            bbox_inches='tight', pad_inches=0.05)

plt.show()
```
